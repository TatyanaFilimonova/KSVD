```python
import plotly.express as px
```

```python
df = px.data.iris()
df
```

Завдання 1

```python
fig = px.scatter(
    df,
    x="sepal_width",
    y="sepal_length",
    color="species",
    size='petal_length',
    hover_data=['petal_width'],
    color_discrete_map={
        "setosa": "#FF6B6B",       # коралово-червоний
        "versicolor": "#4ECDC4",   # бірюзовий
        "virginica": "#FFE66D"     # насичений жовтий
    },
)

fig.update_layout(
    template="plotly_dark",  # темна тема
    plot_bgcolor="#1e1e1e",  # фон графіка
    paper_bgcolor="#1e1e1e", # фон полотна
    font=dict(color="white") # білий шрифт
)

fig.show()
```

```python
fig = px.scatter(
    df,
    x="sepal_width",
    y="sepal_length",
    color='petal_length',
    color_continuous_scale='Viridis'  # приклад градієнта
)
fig.show()
```

```python
fig = px.scatter(df, x="sepal_width", y="sepal_length", color="species", symbol="species")
fig.show()
```

```python
df = px.data.gapminder().query("continent == 'Oceania'")
df.head()
```

image.png

```python
fig = px.line(df, x='lifeExp', y='gdpPercap', color='country')
fig.show()
```

```python
df = px.data.gapminder().query("year == 2007")

fig = px.scatter(
    df, x="gdpPercap", y="lifeExp",
    size="pop", color="continent",
    hover_name="country", log_x=True,
    title="GDP vs Life Expectancy (2007)",
    size_max=60
)
fig.show()
```

```python
# 4.	Зробити візуалізацію зростання тривалості життя  для країни Японія.
df = px.data.gapminder().query("country == 'Japan'")

fig = px.line(
    df, x="year", y="lifeExp",
    title="Life Expectancy in Japan Over Time"
)
fig.show()
```

```python
# 5.	Зробити розподіл довжини пелюстки для різних видів ірису (Тип графіку: Box Plot або Violin Plot)
df = px.data.iris()
df
```

```python
fig = px.violin(
    df, y="petal_length", x="species",
    color="species", box=True, points="all",
    title="Distribution of Petal Length by Species"
)
fig.show()
```
